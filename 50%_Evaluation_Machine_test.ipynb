```python
#imports:
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Question 1:
#Change to dataframe:
df=pd.read_csv("Loan_Eligibility_Prediction.csv")
df
```

```python
#Remove duplicates:
print("No. of duplicate rows:",df.duplicated().sum())
df=df.drop_duplicates()
```

```python
#Question 5
plt.figure(figsize=(8,5))
plt.hist(df["Loan_Amount"], bins=10)
plt.title("Distribution of Loan Amount")
plt.xlabel("Loan Amount")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.countplot(x='Gender',data=df)
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Applicant_Income"], df["Loan_Amount"])
plt.title("Applicant Income vs Loan Amount")
plt.xlabel("Applicant Income")
plt.ylabel("Loan Amount")
plt.show()
```

```python
#Question 8
loan_status = df["Loan_Status"].value_counts()
plt.figure(figsize=(6,6))
plt.pie(loan_status.values,labels=["Approved","Not Approved"],autopct="%.2f%%",startangle=90)
plt.title("Loan Approval Status")
plt.show()
```

```python
plt.hist(df["Loan_Amount"], bins=10)
plt.title("Loan Amount Distribution")
plt.xlabel("Loan Amount")
plt.ylabel("Frequency")
plt.show()
print("The average Loan amount is:",df.Loan_Amount.mean())
```

```python
income = df["Applicant_Income"]
print("Mean:",np.mean(income))
print("Median:",np.median(income))
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="Education",y="Loan_Amount",data=df)
plt.title("Loan Amount by Education")
plt.show()
```

```python
sns.pairplot(df[["Applicant_Income","Coapplicant_Income","Loan_Amount"]])
plt.show()
```

```python
#Question 14
plt.figure(figsize=(8,5))
sns.histplot(data=df, x="Applicant_Income",hue="Loan_Status",bins=30,kde=True)
plt.title("Applicant Income Distribution")
plt.show()
```
