```python
#Question1:
class Parent:
    def show_parent(self):
        print("This is a method from the parent class")
class Child(Parent):
    def show_child(self):
        print("This is an extra method added by the Child class")
obj=Child()

obj.show_parent()
obj.show_child()
```

```python
#Question 2:
class Dog:
    def sound(self):
        print("Dog says: Bow Bow!")

class Cat:
    def sound(self):
        print("Cat says: Meow Meow!")
for animal in (Dog(),Cat()):
    animal.sound()
```

```python
#Question 3:
class Grandparent:
    def method_grandparent(self):
        print("Method of grandparent class (level 1)")
class Parent(Grandparent):
    def method_parent(self):
        print("Method of parent class (level 2)")
class Child(Parent):
        def method_child(self):
            print("Method of Child class (level 3)")
obj=Child()
obj.method_grandparent()
obj.method_parent()
obj.method_child()
```

```python
class Student:
    def __init__(self,name,marks):
        self.name=name
        self.marks=marks

    def display(self):
        print(f"Student Name: {self.name}")
        print(f"Marks: {self.marks}")
s1=Student("Akash",91)
s1.display()
```

```python
class Employee:
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary

    def display_details(self):
        print(f"Student Name: {self.name}")
        print(f"Salary: {self.salary}")

class Manager(Employee):
    def __init__(self,name,salary,department):
        super().__init__(name,salary)
        self.department=department

    def display_manager(self):
        print(f"{self.name} manages the {self.department} department")

m=Manager("Akash",100000,"AI/ML")
m.display_details()
m.display_manager()
        
        
```

```python
#Question 6:
class Shape:
    def	area(self):
        print("Area	formula	depends	on	the	shape")
class	Circle(Shape):
    def	__init__(self,	radius):
        self.radius	=	radius
    def	area(self):
        print(f"Area of Circle={3.14159	*	self.radius	**	2:.2f}")
        
class	Rectangle(Shape):
    def	__init__(self,	length,	breadth):
        self.length	=length
        self.breadth=breadth
    def	area(self):			
        print(f"Area of	Rectangle={self.length	*	self.breadth}")

shapes	=	[Circle(5),	Rectangle(4,	6)]
for	s	in	shapes:
    s.area()
```

```python
#Question 7:
class Arithematic:
    def add(self,a,b):
        return a+b
    def subract(self,a,b):
        return a-b
    def multiply(self,a,b):
        return a*b
    def division(self,a,b):
        return a/b

obj=Arithematic()
print("add=",obj.add(20,4))
print("subract",obj.subract(20,4))
print("multiply=",obj.multiply(20,4))
print("division=",obj.division(20,4))
```
